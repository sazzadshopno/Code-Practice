```python
life_expectancy_data = data_frame['Life_Expectancy']
life_expectancy_data[0]
```

```python
mean = life_expectancy_data.mean()

mean
```

```python
median = life_expectancy_data.median()

median
```

```python
mode = life_expectancy_data.mode()[0]

mode
```

```python
variance = life_expectancy_data.var()

variance
```

```python
standard_deviation = life_expectancy_data.std()

standard_deviation
```

```python
co_variance = (standard_deviation / mean) * 100

co_variance
```

```python
data_range = life_expectancy_data.max() - life_expectancy_data.min()

data_range
```

```python
q1 = life_expectancy_data.quantile(.25)
q3 = life_expectancy_data.quantile(.75)

iqr = q3 - q1

iqr
```

```python
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print(lower_bound, upper_bound)
```

```python
outliers = [x for x in life_expectancy_data if x < lower_bound or x > upper_bound]

outliers
```

```python
# outlier elimination
life_expectancy_data_after_outlier_elimination = [x for x in life_expectancy_data if lower_bound <= x <= upper_bound]

life_expectancy_data_after_outlier_elimination
```

```python
lower_whisker = min([x for x in life_expectancy_data if x >= lower_bound])
upper_whisker = max([x for x in life_expectancy_data if x <= upper_bound])

print(lower_whisker, upper_whisker)
```
